# Sales Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

##### load and understand the dataset

##### Clean the Dataset
1. check for duplicets and remove them
2. check for missing values and handle them/remove them
3. check for invalid datatypes and handle them
4. remove any unnecessary columns

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
amount_mean=df['Amount'].mean()
df['Amount']=df['Amount'].fillna(amount_mean)
```

```python
df['Amount']=df['Amount'].astype('int')
df['Amount'].dtypes 
```

```python
df=df.drop(columns=['Status','unnamed1'])
```

```python
df.info()
```

```python
df['Gender']=df['Gender'].replace({'F':'Female','M':'Male'})
```

##### Data Analysis

```python
plt.figure(figsize=(5,3))
sns.countplot(df,x='Gender')
plt.title('Count of Customers by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

No_of_female customers are  more than male

```python
sales_gen = df.groupby(['Gender'],as_index=False)['Amount'].sum()
sales_gen

```

```python
sns.barplot(x='Gender', y='Amount',data=sales_gen)
plt.title('Total Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.show()
```

Female customers generated higher sales than male customers.

```python
sns.countplot(data=df,x='Age Group',hue='Gender')
plt.title('Customer Count by Age Group and Gender')
plt.xlabel('Age Group')
plt.ylabel('No of Customer')
plt.show()
```

The 26–35 age group has the most male and female customers.

```python
sales_age = df.groupby(['Age Group'],as_index=False)['Amount'].sum()
sales_age
```

```python
sns.barplot(x='Age Group', y='Amount',data=sales_age)
plt.title('Total Sales  by Age Group')
plt.xlabel('Gender')
plt.ylabel('Total Amount')
plt.show()
```

The 26–35 age group contributes the highest sales, followed by the 36–45 age group.

```python
state_order = df.groupby(['State'],as_index=False)['Orders'].sum()
state_order
```

```python
sns.barplot(x='State', y='Orders',data=state_order)
plt.title('Total Orders by State')
plt.xlabel('State')
plt.xticks(rotation=90)
plt.ylabel('Total Orders')
plt.show()
```

```python
state_amount = df.groupby(['State'],as_index=False)['Amount'].sum()
state_amount
```

```python
sns.barplot(x='State', y='Amount',data=state_amount)
plt.title('Total Amount by State')
plt.xlabel('State')
plt.xticks(rotation=90)
plt.ylabel('Total Amount')
plt.show()
```

Uttar Pradesh has the highest number of orders, followed by Maharashtra and Karnataka.

```python
occ_amount = df.groupby(['Occupation'],as_index=False)['Amount'].sum()
occ_amount
```

```python
sns.barplot(x='Occupation', y='Amount',data=occ_amount)
plt.title('Total Sales by Occupation')
plt.xlabel('State')
plt.xticks(rotation=90)
plt.ylabel('Total Amount')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sales_product = df.groupby('Product_Category')['Amount'].sum().sort_values(ascending=False)
sns.barplot(x=sales_product.index, y=sales_product.values)
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

Customers working in the IT Sector contribute the highest sales, followed by Healthcare and Aviation.

```python
top_products = df.groupby('Product_Category')['Orders'].sum().sort_values(ascending=False).head(10)
sns.barplot(x=top_products.index, y=top_products.values)
plt.title('Top 10 Product Categories by Number of Orders')
plt.xlabel('Product Category')
plt.ylabel('Total Orders')
plt.xticks(rotation=45)
plt.show()
```

This chart shows the 10 product categories with the highest total orders, helping identify the most frequently ordered categories.

```python
plt.figure(figsize=(15, 5))
sns.countplot(data=df, x='Product_Category', hue='Gender')
plt.title('Product Category Distribution by Gender')
plt.xlabel('Product Category')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=45)
plt.show()
```

This chart shows which product categories are more popular among male and female customers.

```python
plt.figure(figsize=(10, 5))
top_sales = df.groupby('Product_Category')['Amount'].sum().nlargest(10)
sns.barplot(x=top_sales.values, y=top_sales.index)
plt.title('Top 10 Product Categories by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Category')
plt.show()
```

This chart shows the 10 product categories that generate the highest sales.

1. Female customers generate more sales than male customers.
2. Customers aged 26–35 contribute the highest sales.
3. Uttar Pradesh has the highest sales among the states.
4. Customers in the IT Sector contribute the highest sales among occupations.
5. Maharashtra and Karnataka are among the top states in total sales.
6. Healthcare and Aviation customers contribute significantly to total sales.
7. The Food category contributes the highest sales, followed by Clothing & 
8. The Food category receives the highest number of sales 
9. Clothing & Apparel is one of the top categories contributing to total sales
